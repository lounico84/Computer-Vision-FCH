```python
%load_ext autoreload
%autoreload 2

import importlib
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

import config
importlib.reload(config)

from config import Settings
settings = Settings()
analytics_cfg = settings.analytics

# Teamnamen (falls bei dir in Settings vorhanden)
team_cfg = settings.team_names
team1_name = team_cfg.team1_name
team2_name = team_cfg.team2_name

# Analytics-Module
import analytics.data_loading as dl
import analytics.heatmaps as hm
import analytics.zones as zn
import analytics.possession as pos
import analytics.pass_maps as pm
import analytics.formation as form
import analytics.shots as sh

# Projekt-Root robust finden
PROJECT_ROOT = next(p for p in Path.cwd().parents if (p / "project/Computer-Vision-FCH").exists())

# Pfade robust auflösen
csv_path = Path(settings.paths.frame_events_csv)
if not csv_path.is_absolute():
    csv_path = PROJECT_ROOT / csv_path

pitch_path = Path(settings.paths.pitch_image)
if not pitch_path.is_absolute():
    pitch_path = PROJECT_ROOT / pitch_path

# Laden
df_raw = dl.load_frame_events(csv_path)
df = df_raw.copy()

pitch_img = Image.open(pitch_path)
pitch_length = analytics_cfg.pitch_length
pitch_width  = analytics_cfg.pitch_width
```

```python
from sklearn.cluster import DBSCAN

mask = df["ball_x_m"].notna() & df["ball_y_m"].notna()
points = df.loc[mask, ["ball_x_m", "ball_y_m"]].to_numpy()

if len(points) > 0:
    clust = DBSCAN(eps=1.0, min_samples=200).fit(points)
    labels = clust.labels_

    unique, counts = np.unique(labels, return_counts=True)
    cluster_sizes = dict(zip(unique, counts))

    ARTIFACT_CLUSTERS = [c for c, size in cluster_sizes.items() if size > 300 and c != -1]
    artifact_idx = df.loc[mask].index[np.isin(labels, ARTIFACT_CLUSTERS)]
    df.loc[artifact_idx, ["ball_x_m", "ball_y_m"]] = np.nan

    display(pd.DataFrame([{
        "Artefakt-Cluster": str(ARTIFACT_CLUSTERS),
        "Entfernte Punkte": int(len(artifact_idx))
    }]))
else:
    display(pd.DataFrame([{"Hinweis": "Keine gültigen Meter-Koordinaten gefunden (ball_x_m/ball_y_m leer)."}]))
```

```python
display(HTML(f"<h1>Ballbesitz & Zonen</h1><h2>Zonenverteilung</h2>"))

fig, axes = zn.plot_zone_summary(df, pitch_img, pitch_length, pitch_width)
# Legende im Balkendiagramm anpassen
ax_bar = axes[1]
handles, _ = ax_bar.get_legend_handles_labels()
if len(handles) >= 3:
    ax_bar.legend(handles, ["Gesamt", team1_name, team2_name])
plt.show()

display(HTML("<h2>Rolling Possession</h2>"))
ax_pos = pos.plot_rolling_possession(df, window_sec=10.0)
handles, _ = ax_pos.get_legend_handles_labels()
if len(handles) >= 2:
    ax_pos.legend(handles, [team1_name, team2_name])
ax_pos.set_title(f"Rolling Ball Possession: {team1_name} vs {team2_name}")
plt.show()

page_break()
```

```python
display(HTML(f"<h1>Passanalyse</h1>"))

passes = pm.detect_passes(
    df,
    fps=settings.tracking.fps,
    speed_threshold=5.0,
    min_distance=5.0,
    min_possession_frames=analytics_cfg.pass_min_frames,
)

passes = pm.classify_pass_types(passes, pitch_length=pitch_length)

num_total     = len(passes)
num_clear     = sum(1 for p in passes if p["type"] == "clearance")
num_completed = sum(1 for p in passes if p["type"] == "completed_pass")
num_failed    = sum(1 for p in passes if p["type"] == "failed_pass")

display(pd.DataFrame([{
    "Pässe total": num_total,
    "Angekommen": num_completed,
    "Fehlpässe": num_failed,
    "Klärungen": num_clear
}]))

# Donut-Charts (dein Stil, nur Farben optional)
t1_passes = [p for p in passes if p["team"] == 1]
t2_passes = [p for p in passes if p["team"] == 2]
t1_ok = sum(1 for p in t1_passes if p["completed"])
t2_ok = sum(1 for p in t2_passes if p["completed"])

def draw_pro_donut(ax, total, completed, color_primary, title):
    ax.axis("equal")
    pct = (completed / total) * 100 if total else 0

    ax.pie([100], radius=1.0, colors=["#f0f0f0"], startangle=90,
           wedgeprops=dict(width=0.15, edgecolor="white"))

    if pct > 0:
        ax.pie([pct, 100-pct], radius=1.0, colors=[color_primary, "none"],
               startangle=90, counterclock=False,
               wedgeprops=dict(width=0.15, edgecolor="none"))

    ax.text(0, -0.1, f"{int(pct)}%", ha="center", va="center",
            fontsize=22, fontweight="bold", color="#333")
    ax.text(0, 1.4, f"{total}", ha="center", va="center",
            fontsize=28, fontweight="bold", color="black")
    ax.text(0, 1.85, f"{title}", ha="center", va="center",
            fontsize=12, fontweight="bold", color="gray")

fig, axes = plt.subplots(1, 2, figsize=(10, 5), dpi=120)
plt.suptitle("Erfolgreiche Pässe", fontsize=14, color="gray", y=0.95)
draw_pro_donut(axes[0], len(t1_passes), t1_ok, "#ccb145", team1_name)
draw_pro_donut(axes[1], len(t2_passes), t2_ok, "#b03a42", team2_name)
plt.tight_layout()
plt.show()

display(HTML("<h2>Passmaps</h2>"))
pm.plot_pass_map(passes, team=1, pitch_length=pitch_length, pitch_width=pitch_width,
                 out_path=None, pitch_image_path=str(pitch_path))
plt.show()

pm.plot_pass_map(passes, team=2, pitch_length=pitch_length, pitch_width=pitch_width,
                 out_path=None, pitch_image_path=str(pitch_path))
plt.show()

page_break()
```

```python
%load_ext autoreload
%autoreload 2

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from PIL import Image

# -----------------------------
# Config & Settings
# -----------------------------
import importlib
import config
importlib.reload(config)  # falls du config.py geändert hast

from config import Settings
settings = Settings()
analytics_cfg = settings.analytics

team_cfg = settings.team_names
team1_name = team_cfg.team1_name
team2_name = team_cfg.team2_name

print("Config-CSV-Pfad (relativ/konfiguriert):", settings.paths.frame_events_csv)

# -----------------------------
# Analytics-Module (nur Modul-Imports, keine Funktionen direkt!)
# -----------------------------
import analytics.data_loading as dl
import analytics.heatmaps as hm
import analytics.zones as zn
import analytics.possession as pos
import analytics.pass_maps as pm
import analytics.formation as form

# -----------------------------
# Projekt-Root finden
# -----------------------------
PROJECT_ROOT = next(
    p for p in Path.cwd().parents
    if (p / "project/Computer-Vision-FCH").exists()
)
print("PROJECT_ROOT:", PROJECT_ROOT)

# -----------------------------
# Pfade für CSV & Pitch-Bild robust auflösen
# -----------------------------
csv_path = Path(settings.paths.frame_events_csv)
if not csv_path.is_absolute():
    csv_path = PROJECT_ROOT / csv_path

pitch_path = Path(settings.paths.pitch_image)
if not pitch_path.is_absolute():
    pitch_path = PROJECT_ROOT / pitch_path

print("Verwendete CSV:", csv_path)
print("Verwendetes Pitch-Image:", pitch_path)

# -----------------------------
# Daten laden
# -----------------------------
df_raw = dl.load_frame_events(csv_path)

df = df_raw.copy()

# statische Ball-Artefakte rausfiltern


pitch_img = Image.open(pitch_path)
pitch_length = settings.analytics.pitch_length
pitch_width  = settings.analytics.pitch_width

# Grobe Visualisierung der Ballpunkte in Meterkoordinaten
mask = df["ball_x_m"].notna() & df["ball_y_m"].notna()
plt.figure(figsize=(6, 4))
plt.scatter(
    df.loc[mask, "ball_x_m"],
    df.loc[mask, "ball_y_m"],
    s=1,
    alpha=0.2,
)
plt.axhline(0, color="k", linewidth=0.5)
plt.axvline(0, color="k", linewidth=0.5)
plt.xlabel("x [m]")
plt.ylabel("y [m]")
plt.title("Ballpunkte (raw)")
plt.gca().invert_yaxis()  # falls du oben/unten drehen willst
plt.show()

from sklearn.cluster import DBSCAN
import numpy as np

mask = df["ball_x_m"].notna() & df["ball_y_m"].notna()
points = df.loc[mask, ["ball_x_m", "ball_y_m"]].to_numpy()

# PRÜFUNG: Haben wir überhaupt Daten?
if len(points) == 0:
    print("WARNUNG: Keine gültigen Meter-Koordinaten (ball_x_m, ball_y_m) gefunden!")
    print("DBSCAN und Heatmaps werden leer sein.")
    # Wir machen nichts weiter mit DBSCAN, um den Absturz zu verhindern
else:
    # DBSCAN für dichte Cluster
    clust = DBSCAN(eps=1.0, min_samples=200).fit(points)

    labels = clust.labels_  # -1 = Rauschen, 0,1,2,... = Cluster

    # Clustergrößen zählen
    unique, counts = np.unique(labels, return_counts=True)
    cluster_sizes = dict(zip(unique, counts))

    print("Clustergrössen:", cluster_sizes)

    # Realistische Ballpunkte bewegen sich → keine Cluster > ~300 erlaubt
    ARTIFACT_CLUSTERS = [c for c, size in cluster_sizes.items() if size > 300 and c != -1]
    print("Artefakt-Cluster:", ARTIFACT_CLUSTERS)

    # Index der betroffenen Frames finden
    artifact_idx = df.loc[mask].index[np.isin(labels, ARTIFACT_CLUSTERS)]

    print("Artefaktpunkte entfernt:", len(artifact_idx))

    # Entfernen
    df.loc[artifact_idx, ["ball_x_m", "ball_y_m"]] = np.nan

# -----------------------------
# Basis-Analysen
# -----------------------------
hm.plot_ball_heatmap_on_pitch(df, pitch_img, pitch_length, pitch_width)
plt.title(f"Ball-Heatmap: {team1_name} vs {team2_name}")
plt.show()

fig, axes = hm.plot_team_ball_heatmaps_on_pitch(df, pitch_img, pitch_length, pitch_width)
if len(axes) >= 2:
    axes[0].set_title(f"{team1_name} – Ballbesitz-Heatmap")
    axes[1].set_title(f"{team2_name} – Ballbesitz-Heatmap")
plt.show()

zn.plot_zone_summary(df, pitch_img, pitch_length, pitch_width)
axes[0].set_title(f"Zonen-Verteilung: {team1_name} (Blau) vs {team2_name} (Rot)")

# <--- Balkendiagramm Legende anpassen
ax_bar = axes[1]
handles, _ = ax_bar.get_legend_handles_labels()
if len(handles) >= 3:
    ax_bar.legend(handles, ["Gesamt", team1_name, team2_name])
plt.show()

ax_pos = pos.plot_rolling_possession(df, window_sec=10.0)
handles, _ = ax_pos.get_legend_handles_labels()
if len(handles) >= 2:
    ax_pos.legend(handles, [team1_name, team2_name])
ax_pos.set_title(f"Rolling Ball Possession: {team1_name} vs {team2_name}")
plt.show()

# -----------------------------
# Pässe erkennen
# -----------------------------
passes = pm.detect_passes(
    df,
    fps=settings.tracking.fps,
    speed_threshold=5.0,
    min_distance=5.0,
    min_possession_frames=settings.analytics.pass_min_frames,
)

print("Pässe erkannt:", len(passes))

# Pass-Typen klassifizieren
passes = pm.classify_pass_types(
    passes,
    pitch_length=pitch_length,
)

# Kleine Statistik
num_total     = len(passes)
num_clear     = sum(1 for p in passes if p["type"] == "clearance")
num_completed = sum(1 for p in passes if p["type"] == "completed_pass")
num_failed    = sum(1 for p in passes if p["type"] == "failed_pass")

print("Pass-Übersicht:")
print("  total:     ", num_total)
print("  completed: ", num_completed)
print("  failed:    ", num_failed)
print("  clearances:", num_clear)

# ---------------------------------------------------------
# NEU: Donut-Charts (High Quality "App Style")
# ---------------------------------------------------------
t1_passes = [p for p in passes if p["team"] == 1]
t2_passes = [p for p in passes if p["team"] == 2]

t1_ok = sum(1 for p in t1_passes if p["completed"])
t2_ok = sum(1 for p in t2_passes if p["completed"])

def draw_pro_donut(ax, total, completed, color_primary, title):
    """
    Zeichnet einen Donut-Chart im Stil moderner Fußball-Apps.
    """
    ax.axis('equal') # Wichtig für runde Kreise!
    
    if total == 0:
        pct = 0
    else:
        pct = (completed / total) * 100

    # 1. Hintergrund-Ring (Hellgrau, voller Kreis)
    # Wir zeichnen einen Donut von 0 bis 100% in Grau
    wedges_bg, _ = ax.pie([100], radius=1.0, colors=['#f0f0f0'], startangle=90,
                          wedgeprops=dict(width=0.15, edgecolor='white'))

    # 2. Vordergrund-Ring (Farbe, Anteil)
    # Wenn 100% erreicht sind, ist der Ring voll. Sonst transparent für den Rest.
    if pct > 0:
        # data: [Prozent, Rest]
        # Rest ist transparent (alpha=0)
        wedges_fg, _ = ax.pie([pct, 100-pct], radius=1.0, 
                              colors=[color_primary, 'none'], startangle=90, counterclock=False,
                              wedgeprops=dict(width=0.15, edgecolor='none')) # width muss gleich sein wie oben

    # 3. Text in der Mitte (Prozent)
    ax.text(0, -0.1, f"{int(pct)}%", ha='center', va='center', fontsize=22, fontweight='bold', color='#333333')

    # 4. Text Oben (Gesamtzahl - Groß)
    # Wir nutzen text relative zur Achse oder Daten
    # Position: Etwas über dem Donut (1.0 ist der Rand, also 1.3 darüber)
    ax.text(0, 1.4, f"{total}", ha='center', va='center', fontsize=28, fontweight='bold', color='black')
    
    # 5. Teamname darunter (kleiner)
    ax.text(0, 1.85, f"{title}", ha='center', va='center', fontsize=12, fontweight='bold', color='gray')

# Figure Setup mit hoher DPI für weiche Kanten
fig, axes = plt.subplots(1, 2, figsize=(10, 5), dpi=120)

# Titel über alles
plt.suptitle("Erfolgreiche Pässe", fontsize=14, color='gray', y=0.95)

# Plotten
draw_pro_donut(axes[0], len(t1_passes), t1_ok, '#ccb145', team1_name) # Gold-Ton (ähnlich Real)
draw_pro_donut(axes[1], len(t2_passes), t2_ok, '#b03a42', team2_name) # Rot-Ton (ähnlich Sevilla)

plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# Pass-Maps (wie gehabt)
# ---------------------------------------------------------
# Team 1 Map
pm.plot_pass_map(
    passes, team=1, pitch_length=pitch_length, pitch_width=pitch_width,
    out_path=None, pitch_image_path=pitch_path
)
plt.show()

# Team 2 Map
pm.plot_pass_map(
    passes, team=2, pitch_length=pitch_length, pitch_width=pitch_width,
    out_path=None, pitch_image_path=pitch_path
)
plt.show()


# -----------------------------
# Pass-Maps direkt im Notebook anzeigen
# (create_pass_maps_from_csv ruft intern detect_passes nochmal auf – das ist ok)
# -----------------------------
pm.create_pass_maps_from_csv(
    csv_path=csv_path,
    out_path_team1=None,
    out_path_team2=None,
    pitch_length=analytics_cfg.pitch_length,
    pitch_width=analytics_cfg.pitch_width,
    fps=settings.tracking.fps,
    speed_threshold=analytics_cfg.pass_speed_threshold,
    min_distance=analytics_cfg.pass_min_distance,
    pitch_image_path=pitch_path,
)

print("\n--- Taktische Aufstellung ---")
form.plot_tactical_formation(
    df, 
    pitch_img, 
    pitch_length, 
    pitch_width, 
    team1_name=team_cfg.team1_name, 
    team2_name=team_cfg.team2_name, 
    n_clusters=10
)
plt.show()

# -----------------------------
# 7. Schuss-Statistik & Shot Map
# -----------------------------
import analytics.shots as sh

print("\n--- Schuss-Analyse ---")

# Schüsse erkennen
shots = sh.detect_shots(df, fps=settings.tracking.fps)

# Statistik berechnen
t1_shots = [s for s in shots if s["team"] == 1]
t2_shots = [s for s in shots if s["team"] == 2]

t1_on_target = sum(1 for s in t1_shots if s["on_target"])
t2_on_target = sum(1 for s in t2_shots if s["on_target"])

# Max Speed
t1_max_speed = max([s["speed_kmh"] for s in t1_shots]) if t1_shots else 0
t2_max_speed = max([s["speed_kmh"] for s in t2_shots]) if t2_shots else 0

# DataFrame für schöne Tabelle
stats_data = {
    "Team": [team1_name, team2_name],
    "Schüsse gesamt": [len(t1_shots), len(t2_shots)],
    "Aufs Tor": [t1_on_target, t2_on_target],
    "Genauigkeit %": [
        round(t1_on_target/len(t1_shots)*100, 1) if t1_shots else 0,
        round(t2_on_target/len(t2_shots)*100, 1) if t2_shots else 0
    ],
    "Max km/h": [round(t1_max_speed, 1), round(t2_max_speed, 1)]
}

display(pd.DataFrame(stats_data))

# Plot Shot Map
sh.plot_shot_map(
    shots, pitch_img, pitch_length, pitch_width, 
    team1_name=team1_name, team2_name=team2_name
)
plt.show()
```
